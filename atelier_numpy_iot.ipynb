**Objectif :**

### 1.4 Différence entre une liste Python et un tableau NumPy

**Observation :**

Avec une liste Python :

```text
[24.5, 25.1, 26.3, 27.0, 25.8, 24.5, 25.1, 26.3, 27.0, 25.8]
```

La multiplication par `2` **répète la liste deux fois**.

Avec NumPy :

```text
[49.  50.2 52.6 54.  51.6]
```

La multiplication par `2` **multiplie chaque valeur par 2**.

**Conclusion :**

Une liste Python et un tableau NumPy n'ont donc pas le même comportement avec l'opérateur `*`. NumPy permet notamment d'effectuer directement des opérations numériques sur l'ensemble des valeurs du tableau.

### 7.1 Créer les températures du deuxième bâtiment

Créer aléatoirement la matrice représentant les températures mesurées par les 3 capteurs du deuxième bâtiment durant une journée.

### 7.2 Fusionner les deux bâtiments avec `np.concatenate()`

**Objectif :**
Fusionner les données des 5 capteurs du premier bâtiment et des 3 capteurs du deuxième bâtiment.

Avec `axis=0`, on ajoute les lignes du deuxième bâtiment sous celles du premier.


````python
temperatures_8_capteurs = np.concatenate(
    (temperatures_capteurs_journee, temperatures_batiment_2),
    axis=0
)

print(temperatures_8_capteurs)
````

### 8.1 Calculer les statistiques des températures

**Objectif :**
Calculer le minimum, le maximum, la somme, la moyenne, la médiane, la variance et l'écart-type des températures du premier bâtiment.


````python
minimum = np.min(temperatures_journee)
maximum = np.max(temperatures_journee)
somme = np.sum(temperatures_journee)
moyenne = np.mean(temperatures_journee)
mediane = np.median(temperatures_journee)
variance = np.var(temperatures_journee)
ecart_type = np.std(temperatures_journee)

print("Minimum :", minimum)
print("Maximum :", maximum)
print("Somme :", somme)
print("Moyenne :", moyenne)
print("Médiane :", mediane)
print("Variance :", variance)
print("Écart-type :", ecart_type)
````

### 8.2 Standardiser le tableau

**Objectif :**
Standardiser les températures afin d'obtenir un tableau dont la moyenne est proche de `0` et l'écart-type proche de `1`.

La formule utilisée est :

`z = (x - moyenne) / écart-type`


````python
temperatures_standardisees = (
    (temperatures_journee - np.mean(temperatures_journee))
    / np.std(temperatures_journee)
)

print("Températures standardisées :")
print(temperatures_standardisees)
````

````python
print("Moyenne :", np.mean(temperatures_standardisees))
print("Écart-type :", np.std(temperatures_standardisees))
````

### 8.3 Normaliser les températures

**Objectif :**
Ramener les températures dans l'intervalle `[0, 1]` grâce à la normalisation Min-Max.

La formule utilisée est :

`x_normalise = (x - minimum) / (maximum - minimum)`


````python
temperatures_normalisees = (
    (temperatures_journee - np.min(temperatures_journee))
    / (np.max(temperatures_journee) - np.min(temperatures_journee))
)

print("Températures normalisées :")
print(temperatures_normalisees)
````

````python
print("Minimum :", np.min(temperatures_normalisees))
print("Maximum :", np.max(temperatures_normalisees))
````

### 8.4 Convertir les températures en Fahrenheit

**Objectif :**
Convertir les températures du tableau initial de degrés Celsius en degrés Fahrenheit.

La formule utilisée est :

`F = C × 9/5 + 32`


````python
temperatures_fahrenheit = temperatures_journee * 9 / 5 + 32

print("Températures en Fahrenheit :")
print(temperatures_fahrenheit)
````
