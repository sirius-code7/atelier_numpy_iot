### 1.1 Créer une liste Python

**Objectif :**
Stocker les températures mesurées par un capteur dans une liste Python.

Les températures sont :

`24.5, 25.1, 26.3, 27.0, 25.8`


````python
temperatures_liste = [24.5, 25.1, 26.3, 27.0, 25.8]

print(temperatures_liste)
````

### 1.2 Créer un tableau NumPy

**Objectif :**
Créer un tableau NumPy contenant les mêmes températures que la liste Python précédente.


````python
import numpy as np

temperatures_numpy = np.array([24.5, 25.1, 26.3, 27.0, 25.8])

print(temperatures_numpy)
````

### 1.3 Multiplier les collections par 2

**Objectif :**
Observer le comportement d'une multiplication par 2 sur une liste Python et sur un tableau NumPy.


````python
temperatures_liste_fois_2 = temperatures_liste * 2
temperatures_numpy_fois_2 = temperatures_numpy * 2

print("Liste Python :", temperatures_liste_fois_2)
print("Tableau NumPy :", temperatures_numpy_fois_2)
````

### 1.4 Différence entre une liste Python et un tableau NumPy

**Observation :**

Avec une liste Python :

```text
[24.5, 25.1, 26.3, 27.0, 25.8, 24.5, 25.1, 26.3, 27.0, 25.8]
```

La multiplication par `2` **répète la liste deux fois**.

Avec NumPy :

```text
[49.  50.2 52.6 54.  51.6]
```

La multiplication par `2` **multiplie chaque valeur par 2**.

**Conclusion :**

Une liste Python et un tableau NumPy n'ont donc pas le même comportement avec l'opérateur `*`. NumPy permet notamment d'effectuer directement des opérations numériques sur l'ensemble des valeurs du tableau.

### 5.1 Afficher les températures supérieures à 30 °C

**Objectif :**
Détecter les températures dépassant `30 °C`.

Le filtrage booléen permet de sélectionner uniquement les valeurs qui respectent une condition.


````python
temperatures_superieures_30 = temperatures_journee[temperatures_journee > 30]

print("Températures supérieures à 30 °C :", temperatures_superieures_30)
````

### 5.2 Afficher les températures normales

**Objectif :**
Afficher les températures comprises entre `20 °C` et `30 °C`.

On utilise deux conditions combinées avec `&`.


````python
temperatures_normales = temperatures_journee[
    (temperatures_journee >= 20) & (temperatures_journee <= 30)
]

print("Températures normales :", temperatures_normales)
````

### 5.3 Déterminer les températures anormales

**Objectif :**
Créer un tableau booléen indiquant quelles températures sont supérieures à `35 °C`.

`True` indique une anomalie et `False` une température normale.


````python
anomalies = temperatures_journee > 35

print("Anomalies :", anomalies)
````

### 5.4 Déterminer le nombre d'anomalies

**Objectif :**
Compter le nombre de températures considérées comme anormales à partir du tableau booléen `anomalies`.

Dans NumPy, les valeurs `True` sont comptées comme `1` et les valeurs `False` comme `0`.


````python
nombre_anomalies = np.sum(anomalies)

print("Nombre de températures anormales :", nombre_anomalies)
````

### 6.1 Créer la matrice des températures

**Objectif :**
Créer aléatoirement la matrice représentant les températures mesurées par 5 capteurs pendant une journée.

Chaque ligne représente un capteur et chaque colonne représente une heure.


````python
temperatures_capteurs_journee = np.random.uniform(20, 30, (5, 24))

print(temperatures_capteurs_journee)
````

### 6.2 Aplatir la matrice

**Objectif :**
Transformer la matrice des températures en un tableau unidimensionnel.

La méthode `flatten()` crée une copie aplatie du tableau.


````python
temperatures_aplaties = temperatures_capteurs_journee.flatten()

print(temperatures_aplaties)
````

### 6.3 Déterminer la transposée de la matrice

**Objectif :**
Inverser les lignes et les colonnes de la matrice.

La transposée transforme une matrice de forme `(5, 24)` en une matrice de forme `(24, 5)`.


````python
temperatures_transposees = temperatures_capteurs_journee.T

print(temperatures_transposees)
````
